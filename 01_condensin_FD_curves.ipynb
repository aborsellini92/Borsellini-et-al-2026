# ALL ON ONE GRAPH:

```python
import os
import lumicks.pylake as lk
import matplotlib.pyplot as plt
import itertools

# Distinguishable color palette (Tableau 10, has 10 good distinct colors)
colors = itertools.cycle(plt.get_cmap("tab10").colors)

plt.figure(figsize=(8,6))

for file_name in h5_files:
    start = '1s'
    end = '4s'
    
    file_path = os.path.join(folder_path, file_name)
    file = lk.File(file_path)

    distance_selection = file['Distance']['Distance 1'][start:end].data
    force_selection = file['Force LF']['Force 2x'][start:end].data

    # assign next color from cycle
    color = next(colors)

    # label curve by file name (without extension for clarity)
    label = os.path.splitext(file_name)[0]

    plt.scatter(distance_selection, force_selection, 
                s=1, label=label, color=color)

plt.ylabel("Force (pN)")
plt.xlabel("Distance ($\mu$m)")

plt.axhline(0, color="black", linewidth=0.8)
plt.xlim(1, 18)
plt.ylim(0, 70)

# ✅ Correctly closed legend call
plt.legend(markerscale=4, fontsize=8)  

plt.tight_layout()
plt.show()

```

# Export graphs in excel

```python
#Export graphs in excel

import os
import csv
import lumicks.pylake as lk
import matplotlib.pyplot as plt

# Define output file
output_file = os.path.join(folder_path, "exported_data.csv")

# Dictionary to store data from all files
data_dict = {}

# Process each file
for file_name in h5_files:
    start = '0s'
    end = '4s'
    
    file_path = os.path.join(folder_path, file_name)
    file = lk.File(file_path)

    distance_selection = file['Distance']['Distance 1'][start:end].data
    force_selection = file['Force LF']['Force 2x'][start:end].data

    # Store data in a dictionary
    for d, f in zip(distance_selection, force_selection):
        if d not in data_dict:
            data_dict[d] = {}  # Create a sub-dictionary for this distance value
        data_dict[d][file_name] = f  # Store force value under corresponding file

    # Plot the data
    plt.scatter(distance_selection, force_selection, s=1, label=file_name)

# Sort the data by Distance (ascending order)
sorted_distances = sorted(data_dict.keys())

# Write data to CSV
with open(output_file, mode='w', newline='') as file:
    writer = csv.writer(file)

    # Create header: "Distance" + one column per file
    header = ["Distance (µm)"] + h5_files
    writer.writerow(header)

    # Write each row
    for d in sorted_distances:
        row = [d] + [data_dict[d].get(file_name, "") for file_name in h5_files]
        writer.writerow(row)

# Plot settings
plt.ylabel('Force (pN)')
plt.xlabel('Distance ($\mu$m)')
plt.axhline(0, color='black')
plt.xlim(1, 18)
plt.ylim(0, 70)
plt.legend()
plt.show()

print(f"Data saved to {output_file}")
```
